---
## **Enunciado**

Reconocimiento de señales de tráfico en imágenes digitales sabiendo que se identifican por
color (azul, rojo y amarillo) y forma (círculo, cuadrado, octógono, triángulo y triángulo inverso)
conforme a lo mostrado en la siguiente tabla:

imagen.png

**Se pide**: un programa en python que dada una imagen en la que aparezca una señal de tráfico,
nos diga qué señal concreta es, conforme a lo indicado en la tabla anterior

**Ejemplos**:

imagen.png


**Optativo**: quien quiera profundizar en el desarrollo de esta práctica (no es de entrega obligatoria), puede emplear alguna otra característica (además del color y la forma) como, por ejemplo: el reconocimiento del texto (caracteres y dígitos) que hay en el interior de una señal de tráfico.

En nuestro caso, el empleo de reconocimiento de texto nos permite incluir las siguientes precisiones:
- una señal de color azul y forma circular, si el texto que contiene es numerico, se convierte en una señal de recomendación que nos indica la velocidad mínima a la que se recomienda circular

- una señal de color azul y forma cuadrada, si el texto que contiene es numerico, se convierte en una señal de recomendación que nos indica la velocidad máxima a la que se recomienda circular (y si no hay texto entonces se convierte en una señal de paso de peatones)

- una señal de color rojo y forma circular, si el texto que contiene es numerico, se convierte en una señal de prohibido que nos indica la velocidad máxima a la que se puede circular (de forma similar obtenemos lo mismo para las señales amarillas de formas circulares, octogonales, triángulares y triángulares inversas)


**Nota**: para el reconocimiento del texto (caracteres y dígitos) que hay en el interior de una señal de tráfico se puede emplear cualquier biblioteca de funciones en python que posibilite manejar un OCR.
